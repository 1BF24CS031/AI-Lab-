```python
import random

def get_attacking_pairs(state):
    """Calculates the number of attacking pairs of queens."""
    n = len(state)
    attacks = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Same column or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == j - i:
                attacks += 1
    return attacks

def get_neighbors(state):
    """Generates all possible neighbor states by moving one queen within its row."""
    neighbors = []
    n = len(state)
    for row in range(n):
        for col in range(n):
            if state[row] != col:
                neighbor = list(state)
                neighbor[row] = col
                neighbors.append(neighbor)
    return neighbors

def print_board(state):
    """Prints the board in a user-friendly text format."""
    n = len(state)
    for row in range(n):
        line = ["Q" if state[row] == col else "." for col in range(n)]
        print(" ".join(line))
    print()

def hill_climbing_n_queens_verbose(n, max_restarts=100):
    """Solves N-Queens using Hill Climbing, printing the board states at each transition."""
    for restart in range(max_restarts):
        print(f"\n=== Restart #{restart} ===")
        current_state = [random.randint(0, n - 1) for _ in range(n)]
        current_value = get_attacking_pairs(current_state)
        print(f"Initial State: {current_state} (Attacking Pairs: {current_value})")
        print_board(current_state)

        step = 0
        while True:
            if current_value == 0:
                print(f"Goal state reached!")
                return current_state, restart

            neighbors = get_neighbors(current_state)
            best_neighbor = None
            best_value = current_value

            for neighbor in neighbors:
                val = get_attacking_pairs(neighbor)
                if val < best_value:
                    best_value = val
                    best_neighbor = neighbor

            if best_value >= current_value:
                print(f"Local optimum hit at State: {current_state} (Attacking Pairs: {current_value}). Triggering restart.\n")
                break

            step += 1
            print(f"  Step {step}: Moved to State: {best_neighbor} (Attacking Pairs: {best_value})")
            print_board(best_neighbor)
            current_state = best_neighbor
            current_value = best_value

    return None, max_restarts

# Ask for user input and solve with verbose output
try:
    N = int(input("Enter the number of queens (N): "))
    if N < 4 and N != 1:
        print("Note: No solutions exist for N = 2 or 3.")

    print(f"Searching for a solution for {N}-Queens with verbose steps...")
    solution, restarts = hill_climbing_n_queens_verbose(N)

    if solution:
        print(f"\nSuccess! Solution found with {restarts} random restarts:")
        print(f"Final State: {solution}\n")
        print_board(solution)
    else:
        print("Could not find a solution within the restart limit.")
except ValueError:
    print("Please enter a valid integer.")
```
